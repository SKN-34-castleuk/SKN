```python
# 항등 함수

import numpy as np

def identity_function(x):
    return x

x = np.array([1.0, 2.0, 3.0])
print(identity_function(x))
```

회귀문제도 신경망 내부에서 이미 은닉층을 거치면서 비선형 변환이 이루어진 상태이므로,  
출력층에서는 그냥 나오는 출력값 그대로 사용한다.

```python
# 시그모이드 함수
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.array([0.0, 2.0, -1.0])
print(sigmoid(x))
```

Sigmoid 함수로 나온 예츨확률을 임계값을 기준으로 예측 결과로서 결정  
출력값 >= 0.5 -> 클래스 1  
출력값 < 0.5 -> 클래스 0    

- 활성화 함수로서 사용하지 않는 이유  
출력이 0 또는 1에 가까워지면 기울기가 거의 0에 수렴되어 학습이 거의 되지 않아  
출력층에서만 사용한다.  

손실함수로 BCE와 함께 사용

```python
# 소프트 맥스 함수

def softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

x = np.array([2.0, 1.0, 0.1])
print(softmax(x))
```

- 소프트맥스 함수 구현 시 주의점
    - 소프트맥스 함수는 지수 연산을 포함하기 때문에 오버플로우(Overflow) 문제가 발생할 수 있다.
    - 이를 방지하기 위해 입력에서 최대값을 빼는 기법을 적용한다.

```python
def stable_softmax(x):
    x_max = np.max(x)
    exp_x = np.exp(x - x_max)  # 최대값을 빼서 안정적으로 변환
    return exp_x / np.sum(exp_x)

x = np.array([1000, 1001, 1002])  # 큰 값에서도 안정적으로 동작
print(stable_softmax(x))
```
