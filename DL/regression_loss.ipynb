```python
import torch
import torch.nn.functional as F
```

```python
y_true = torch.tensor([3.0,5.0,7.0]) # 실제값
y_pred = torch.tensor([2.5,5.5,10.0]) # 예측값
```

```python
mse_loss = F.mse_loss(y_true,y_pred)        # MSE : 평균제곱오차
mae_loss = F.l1_loss(y_true,y_pred)         # MAE : 평균절대오차
huber_loss = F.huber_loss(y_true,y_pred)    # Huber 손실

print(f"{mse_loss.item():.4f}")
print(f"{mae_loss.item():.4f}")
print(f"{huber_loss.item():.4f}")
```

```python
y_pred_outlier = torch.tensor([10,245,23])

mse_loss = F.mse_loss(y_true,y_pred_outlier)        # MSE : 평균제곱오차
mae_loss = F.l1_loss(y_true,y_pred_outlier)         # MAE : 평균절대오차
huber_loss = F.huber_loss(y_true,y_pred_outlier)    # Huber 손실

print(f"{mse_loss.item():.4f}")
print(f"{mae_loss.item():.4f}")
print(f"{huber_loss.item():.4f}")
```

MSELoss는 제곱연 산을 하기 떄문에 크 오차가 발생하면 손실값이 크게 증가한다.  
이상치가 많은 데이터의 경우 MAELoss나 HuberLoss가 더 안정적으로 동작한다.
