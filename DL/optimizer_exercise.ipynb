```python
import numpy as np

def sgd_update(w, grad, lr=0.01):
    return w - lr * grad

# 가중치와 기울기
w = np.array([0.5, -0.3])       # 현재 가중치
grad = np.array([0.2, -0.4])    # 현재 기울기 벡터

# SGD 업데이트 적용
w_new = sgd_update(w, grad)         # SGD로 새 가중치 계산
print("Updated weights:", w_new)
```

```python
def momentum_update(w, v, grad, lr=0.01, momentum=0.9):
    v = momentum * v - lr * grad
    w += v
    return w, v

# 초기값 설정
w = np.array([0.5, -0.3])
v = np.array([0.0, 0.0])
grad = np.array([0.2, -0.4])

# 모멘텀 업데이트 적용
w_new, v_new = momentum_update(w, v, grad)
print(f"Updated weights: {w_new}, 업데이트 된 속도 : {v_new}")
```

```python
def adagrad_update(w, grad, h, lr=0.01, epsilon=1e-8):
    h += grad ** 2
    w -= (lr / (np.sqrt(h) + epsilon)) * grad
    return w, h

# 초기값 설정
w = np.array([0.5, -0.3])
h = np.array([0.0, 0.0])        # 희소성 판단 (기울기 제곱 누적)
grad = np.array([0.2, -0.4])    

# AdaGrad 업데이트 적용
w_new, h_new = adagrad_update(w, grad, h)
print("Updated weights:", w_new,h_new)
```

```python
def rmsprop_update(w, h, grad, lr=0.01, beta=0.9, epsilon=1e-8):
    h = beta * h + (1 - beta) * (grad ** 2)
    w -= (lr / (np.sqrt(h) + epsilon)) * grad
    return w, h
```

```python
def adam_update(w, m, v, grad, t, lr=0.01, beta1=0.9, beta2=0.999, epsilon=1e-8):
    m = beta1 * m + (1 - beta1) * grad          # 1차 (모멘텀)
    v = beta2 * v + (1 - beta2) * (grad ** 2)   # 2차 (RMSProp)
    m_hat = m / (1 - beta1 ** t)    # 1차 모먼트 초기값 보정
    v_hat = v / (1 - beta2 ** t)    # 2차 모먼트 초기값 보정
    w -= (lr / (np.sqrt(v_hat) + epsilon)) * m_hat  # Adam 옵티마이저 (적응적 스텝)
    return w, m, v
```
