```python
import torch

x = torch.FloatTensor([[1,2],
                      [3,4]]).requires_grad_(True)
```

requires_grad_(True) : 이 텐서에 대해서 미분값을 구하겠다.(일반적인 학습에서는 사용안함.  
기울기 분석용)

```python
# x -> x1, x2 -> x3 -> y 연산 트리 생성

x1 = x +2
x2 = x-2
x3 = x1 * x2
y = x3.sum()

print(x1)
print(x2)
print(x3)
print(y)
```

```python
y.backward() # y값을 기준으로 계산 그래프를 따라 연결된 모든 텐서의 기울기르 자동으로 계산
```

```python
print(x.grad)   # y에 대해 계산된 x의 기울기(∂y/∂x)
```

y = ∑(x+2)(x-2) 모두 합치면 y= ∑(x^2-4)

∂/∂x(-4) = 0     
=> -4는 미분 결과에 영향을 주지 않음

제곱항은 $ \frac{\partial}{\partial x_i}(x_i^2) = 2x_i $

각 원소에 대해서
$ \frac{\partial y}{\partial x_i} = 2x_i $  

벡터 형태로 보면
$ \frac{\partial y}{\partial x} = 2x $  

결론 적으로 x.grad 결과는 2x 형태로 나오게 된다.

```python
print(x)        # 텐서 값은 변경 없음 . 역전파는 기울기를 계산만하고, 수정하지 않음. (기울기 추적 대상)
```

```python
x3.numpy() # numpy는 autograd 상태를 모르기때문에 바로 numpy형태로 변환은 되지않음
```

```python
x3.detach_().numpy()    # x3를 계산 그래프에서 분리(autograd 추적 제거)하고 numpy 배열로 반환
```

detach() : 복사본을 만들면서 분리 (권장)  
detach_() : 원본자체를 분리(detach() + inplace 옵션)
