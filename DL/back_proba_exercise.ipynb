```python
import numpy as np

# 활성화 함수 (시그모이드)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 활성화 함수의 미분
def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))

# 입력 데이터
x = np.array([0.5, 0.8])
# 실제 출력값
y_true = np.array([1])

# 가중치 초기화
w = np.array([0.2, 0.4])

# 순전파
z = np.dot(x, w)  # 선형 변환
a = sigmoid(z)  # 활성화 함수 적용

# 오차 계산
loss = 0.5 * (y_true - a) ** 2

# 역전파 (기울기 계산)
delta = (a - y_true) * sigmoid_derivative(z)    # 출력 오차* 활성화 함수 미분
grad_w = delta * x  # 연쇄법칙 적용 (기울기 계산)

# 가중치 업데이트
learning_rate = 0.1
w -= learning_rate * grad_w # 경사 하강법

print("업데이트된 가중치:", w)
```

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 입력 데이터
x = np.array([0.5, 0.8])
# 가중치
w = np.array(
    [[0.1, 0.3],    # 입력 1이 은닉 노드로 가는 가중치
     [0.2, 0.4]])   # 입력 2가 은닉 노드로 가는 가중치
# 편향
b = np.array([0.1, 0.2])    # 은닉 노드에 더해지는 값

# 은닉층 계산 (국소적 계산)
z = np.dot(x, w) + b
a = sigmoid(z)

print("은닉층 출력:", a)
```

```python
import numpy as np

def forward(x):
    y = x ** 2  # 예제 함수: y = x^2
    z = 2 * y   # z = 2y
    return z

def backward(x):
    dy_dx = 2 * x  # y = x^2의 미분
    dz_dy = 2      # z = 2y의 미분
    dz_dx = dz_dy * dy_dx  # 연쇄법칙 적용
    return dz_dx

x = 3.0
print("순전파 결과:", forward(x))
print("역전파 결과 (기울기):", backward(x)) # 입력 x에 대한 z의 기울기
```

```python
import numpy as np

def f(x):
    return x**2  # 예제 함수: y = x^2

def numerical_gradient(f, x):
    h = 1e-4  # 작은 값
    return (f(x + h) - f(x - h)) / (2 * h)  # 중심 차분법 사용

def analytical_gradient(x):
    return 2 * x  # y = x^2의 미분 결과

x = 3.0
num_grad = numerical_gradient(f, x)
ana_grad = analytical_gradient(x)

print("수치 미분 결과:", num_grad)
print("오차역전파법 결과:", ana_grad)
print("차이:", abs(num_grad - ana_grad))
```
