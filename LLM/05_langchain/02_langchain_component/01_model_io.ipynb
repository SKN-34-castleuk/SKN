# Model  

## OpenAI

```python
from langchain_openai import ChatOpenAI     # OpenAI 채팅 모델을 Langchain에서 쓰는 래퍼 클래스

llm = ChatOpenAI(model = 'gpt-5.6-luna')    # LLM 객체 생성
# 사용자 입력으로 모델 호출 -> AIMessage(content) 반환
print(llm.invoke('전주 여행가려고 해. 계획 만들어 줘').content) 
```

```python
from langchain.chat_models import init_chat_model       # 추상화된 llm모델 객체 생성

# 객체 생성 : 모델 + 하이퍼파라미터 설정
llm = init_chat_model(model='gpt-5.6-luna',temperature=1,top_p=1)
ai_message = llm.invoke('속초 당일치기 여행계획 만들어 줘.')    # 프롬프트 -> AIMessage 반환

print(ai_message.content)
```

- 추상화된 채팅 LLM : init_chat_model() -> 특정 회사 (OpenAI / Anthropic / HuggingFace)의 클래스에 직접 묶이지 않고, 공통 인터페이스로 같은 방식(invoke, streeam 등)으로 쓰게 만들어준다.  
- 즉, init_chat_model()은 OpenAI 객체를 만드는 방식이 아니라, 여러 채팅 LLM을 공통 규격으로 감싸서 (추상화화여) 동일하게 쓸 수 있도록 만들어 주는 팩토리 함수이다.

## HuggingFace


```python
from langchain.chat_models import init_chat_model       # 추상화된 llm모델 객체 생성

# 객체 생성 : 모델 + 하이퍼파라미터 설정
llm = init_chat_model(model='huggingface:Qwen/Qwen2.5-0.5B-Instruct', temperature=1)
ai_message = llm.invoke('속초 당일치기 여행계획 만들어 줘.')    # 프롬프트 -> AIMessage 반환

print(ai_message.content)
```

## 여러 모델 후보군 비교

```python
models = [
    init_chat_model('openai:gpt-5.6-luna'),
    init_chat_model('openai:gpt-4.1-mini'),
    init_chat_model('huggingface:Qwen/Qwen2.5-0.5B-Instruct')
]

user_prompt = '취업 어떻게 하지?'

for model in models:
    # model객체에서 modle_name 속성이 있으면 model_name, 없으면 model_id를 사용
    model_name = model.model_name if hasattr(model,'model_name') else model.model_id  
    print(f"==={model_name}")
    ai_message = model.invoke(user_prompt)
    print(ai_message.content)
    print()
```

### Prompt

```python
llm = init_chat_model('openai:gpt-5.6-luna')

messages = [
    ('system', '당신은 친절한 챗봇입니다.'),        # 시스템 프롬프트 : 역할 (페르소나) / 규칙
    ('user', '랭체인/랭그래프를 초보자도 알 수 있을정도로 간단하게 설명해주세요.'),
]

ai_message = llm.invoke(messages)
print(ai_message.content)
```

```python
# PromptTemplate으로 프롬프트를 템플릿화해서 상품별 문구를 만들고 LLM을 생성
from langchain_core.prompts import PromptTemplate

# 변수 {product}를 포함한 템플릿 생성
prompt = PromptTemplate.from_template('{product}를 홍보하기 위한 참신한 문구를 생성해줘.')

print(prompt)

print(prompt.format(product = '전기차'))        # 변수를 채운 완성된 프롬프트 문자열
print(prompt.format(product = '카메라'))

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(prompt.format(product = '부동산관련 웹페이지'))
print(ai_message.content)
```

```python
# PromptTemplate으로 프롬프트를 템플릿화해서 상품별 문구를 만들고 LLM을 생성
from langchain_core.prompts import ChatPromptTemplate

# 여러 메시지(role, content)를 템플릿으로 구성
chat_prompt = ChatPromptTemplate.from_messages([
    ('system' ,'당신은 {domain}분야의 전문가입니다. 친절히 답해주세요.'),
    ('user','{question} 한글로 답해주세요.')
])

print(chat_prompt.format(domain = 'IT', question = 'langsmith 사용법')) # 변수를 채운 문자열


llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(chat_prompt.format(domain = 'IT', question = 'langsmith 사용법'))
print(ai_message.content)
```

## Outpur Parser  

객체의 출력에서 실제 값부분을 가져와 후처리 작업 담당

```python
# CommaSeparatedListOutputParser() : 콤마 구분 문자열 -> 리스트 변환 파서
from langchain_core.output_parsers import CommaSeparatedListOutputParser

parser = CommaSeparatedListOutputParser()
model_output = '달걍, 고구마, 대파, 쪽파'
parser.parse(model_output)
```

```python
format_instruction = parser.get_format_instructions()
print(format_instruction)
```

```python
prompt = PromptTemplate(
    template='{subject} 팀 {n}개를 작성해주세요. \n(지시사항 : {format_instruction})',
    input_variables= ['subject','n'],
    partial_variables = {"format_instruction": format_instruction}
)

print(prompt)
print(prompt.format(subject = 'kbo', n=7))
```

```python
input = prompt.invoke({'subject': 'kbo', 'n': 7})
print(input)

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(input)
print(ai_message.content)
```

```python
# StrOutoutParser : 모델 답변을 가공없이 문자열로 받아서 출력

from langchain_core.output_parsers import StrOutputParser   # AIMessage -> str(content) 파서

parser = StrOutputParser()
chain = llm | parser    # LLM출력 -> 문자열로 파싱
chain.invoke('아 너무 덥다! 여름에 먹을만한 제철과일은 뭐야?')
```

```python
# JsonOutputParser : 모델의 json 문자열 응답을 파싱하여 dict로 출력

from langchain_core.output_parsers import JsonOutputParser

parser = JsonOutputParser()
format_instruction = parser.get_format_instructions()
print(format_instruction)

prompt = PromptTemplate(
    template='{subject} 분야에서 유명한 한국어 책{n}권 정보를 제공해주세요. \n(지시사항 : {format_instruction})',
    input_variables= ['subject','n'],   # 사용자가 채울 변수들
    partial_variables = {"format_instruction": format_instruction}  # 고정으로 주입할 변수
)

chain = prompt | llm | parser 
result  = chain.invoke({'subject':'회빙환','n' : 10})
result
```

```python
# PydanticOutputParser : 형식을 강제해서 구조화된 형태로 받는 파서
from pydantic import BaseModel   # 데이터 스키마(모델) 정의용
from langchain_core.output_parsers import PydanticOutputParser  # LLM 출력 -> Pydantic 객체로 검증/변환
from typing import List # 리스트 타입 힌트용

# 책 1권마다의 스키마 정의
class Book(BaseModel):
    title: str
    author : str
    publisher : str
    year : int = None   # 없어도 상관없음
    description : str

# 책 여러권을 담는 스키마 정의
class Booklist(BaseModel):
    books : List[Book]

parser = PydanticOutputParser(pydantic_object= Booklist)    # 출력이 BookList 형태가 되ㅣ도록 파서 생성
format_instruction = parser.get_format_instructions()       # 스키마대로 출력하도록 만드는 지시문 생성
print(format_instruction)

prompt = PromptTemplate(
    # subject/ n+ 지시사항을 포함한 템플릿
    template='{subject} 분야에서 유명한 한국어 책{n}권 정보를 제공해주세요. \n(지시사항 : {format_instruction})',
    input_variables= ['subject','n'],   # 사용자가 채울 변수들
    partial_variables = {"format_instruction": format_instruction}  # 고정으로 주입할 변수
)

chain = prompt | llm | parser       # 체인 : prompt -> LLM -> PYdantic 파서
result  = chain.invoke({'subject':'인문소설','n' : 5})  # chain 실행
result
```

- Book / Booklist는 pydantic.BaseModel을 상속한 스키마(데이터 모델) 이다.  
- result는 그 스키마로 만들어진 pydantic 객체(인스턴스) 
- Pydantic 을 사용하면    
    - 필드가 있는지 검사
    - 타임 검사/변환  
    - 구조가 맞는지 검사
- dictionary와의 차이점
    - dict: 아무키나 들어갈 수 있고, 타입도 상관없다.
    - Pydantic 객체 : 정해진 키/타입만 허용.틀리면 검증단계에서 수정해줌

```python
# with_structured_output() : Pydantic 스키마를 값으로 받아 해당 형태의 구조화된 결과를 바로 반환
prompt = PromptTemplate(
    # subject/ n+ 지시사항을 포함한 템플릿
    template='{subject} 분야에서 유명한 한국어 책{n}권 정보를 제공해주세요.',
    input_variables= ['subject','n']   # 사용자가 채울 변수들
)

chain = prompt | llm.with_structured_output(Booklist)   # LLM 출력이 BookList 스키마를 따르도록 래핑
result  = chain.invoke({'subject':'인공지능','n' : 5})      # chain 실행
result
```

```python
for book in result.books:
    print(f"책 제목 : {book.title}")
    print(f"책 저자 : {book.author}")
    print(f"출판사 : {book.publisher}")
    print(f"출판연도 : {book.year}")
    print(f"설명: {book.description}")
    print()
```
