```python
import ollama

# api 형식으로 ollama 실행
response = ollama.chat(
    model='llama3', 
    messages=[{'role': 'user', 'content': '안녕하세요!'}]
)

print(response)
```

```python
# 텍스트를 입력받아 로컬LLM의 답변을 생성하는 함수
def ollama_chat(input_text):
    return ollama.chat(
        model='llama3', 
        messages=[{'role': 'user', 'content': input_text}]
    ).message.content
```

```python
ollama_chat('파이썬으로 퀵소트를 구현해줘.')
```

```python
ollama_chat('다음 텍스트를 요약해줘: ...')
```
