```python
# 한국어 -> 영어 번역
translator = pipeline('translation', model= 'Helsinki-NLP/opus-mt-ko-en')
translator
```

```python
translator('와 이제 마지막 시간이다!')
```

```python
translator([
    '와 이제 마지막 시간이다!',
    '이제 수업을 열심히 들어보자',
    '집도 열심히 갑시다!',
    '내일은 일찍와',
])
```

```python
sentiment_clf = pipeline('sentiment-analysis')
```

```python
sentiment_clf("I'm very happy!")
```

```python
ko_sentiment_clf = pipeline('sentiment-analysis', model = 'sangrimlee/bert-base-multilingual-cased-nsmc')
```

```python
ko_sentiment_clf([
    '오늘 배운거는 너무 쉬운데?',
    '그래서 눈물이 자꾸 나',
    '힘내자'
])
```

```python
# Zero-Shopt Classification

zero_shot_clf = pipeline('zero-shot-classification')
```

```python
zero_shot_clf(
    'This is a course about the transformers library',
    candidate_labels = ['education', 'politics','business']
)
```

```python
zero_shot_clf(
    "It has a poison and it's very dangerous",
    candidate_labels = ['captbara', 'tiger','snake']
)
```

```python
ko_zero_shot_clf = pipeline('zero-shot-classification', model='joeddav/xlm-roberta-large-xnli')
```

```python
ko_text = '2026년 당신은 얼마나 잘 살고있나요?'
candidates = ['사회', '경제', '건강', '정치']

ko_zero_shot_clf(ko_text, candidates)
```

```python
# 텍스트 생성
text_generator = pipeline('text-generation', model = 'distilgpt2')
```

```python
text_generator(
    'In the course, we will teach you hot to',
    max_length =30,
    num_return_sequences = 2
)
```

```python
ko_text_generator = pipeline('text-generation', model = 'skt/ko-gpt-trinity-1.2B-v0.5')
```

```python
ko_text_generator('오늘 저녁에는')
```

```python
# Fill Mask
unmasker = pipeline('fill-mask')
```

```python
unmasker('This course will teach you all about <mask> model', top_k = 3)
```

```python
# NER (객체명 인식)
ner = pipeline(
    'ner', 
    model = 'dslim/bert-base-NER', 
    grouped_entities = True    # 연속된 동일 개체는 하나로 묶어서 반환
)
```

```python
ner('My name is Capybara and I work at Playdata in seoul')
```

```python
# Q&A
qna = pipeline('question-answering')
```

```python
qna(
    question = 'Where do i work?',
    context = 'My name is Capybara and I work at Playdata in seoul'
)
```

```python
ko_qna = pipeline('question-answering', model = 'klue/roberta-base')
```

```python
ko_qna(
    question="1931년 어떤 진단을 받았나요?",  # 문맥에서 찾고자 하는 질문
    context="""그러던 1931년, 그는 갑작스럽게 폐결핵을 진단받았다. 이상은 하루에 담배를 50개피 이상 피는 것을 자신의 일과라고 표현했을 정도로 엄청난 골초였는데, 그 때문인지 병세는 날이 갈수록 악화되어 담당의가 그의 폐를 확인하고는 형체도 안 보인다며 혀를 내둘렀을 정도였다. 결국 1933년부터는 각혈까지 시작되었고, 건축기사 일을 지속하기 어렵다고 판단한 이상은 조선총독부에서 퇴사하고 황해도에 있는 배천 온천으로 요양을 간다."""  # 답변을 추출할 문맥
)
```
