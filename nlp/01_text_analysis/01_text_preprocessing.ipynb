```python
from nltk.tokenize import word_tokenize, sent_tokenize # 단어/문장 토큰화 도구

text = "NLTK is a powerful library for NLP!!!!!"
word_tokenize(text) # 단어 토큰화
```

```python
text = "NLTK is a powerful library for NLP! NLTK is a powerful library for NLP. NLTK is a powerful library for NLP?"
sent_tokenize(text) # 문장 토큰화
```

```python
def tokenize_text(text):
    sentences = sent_tokenize(text) # 문장토큰화
    return [word_tokenize(sentence) for sentence  in sentences]

tokenize_text(text)
```

```python
# NLTK N-gram(바이그램/트라이그램) 생성
from nltk import ngrams

text = "The Martrix is everywhere its all arounds us, here even in this room"

tokens = word_tokenize(text)   # 단어 토큰화
print(tokens)

bigram = ngrams(tokens, 2)      #2-gram(바이그램) 생성
print([token for token in bigram])

trigram = ngrams(tokens, 3)     #3-gram(트라이그램) 생성
print([token for token in trigram])
```

```python
text = "The Martrix is everywhere its all arounds us, here even in this room"
stopwords_list = stopwords.words('english') # 영어 불용어 리스트

tokens =[]
for word in word_tokenize(text):    # 단어 토큰화
    word = word.lower()             # 소문자 변환
    if word not in stopwords_list:  # 불용어 제외
        tokens.append(word)

tokens
```

### 특성 벡터화
- BOW > CounVectorizer

```python
text1 = 'The Matrix is everywhere its all around us, here even in this room. \
You can see it out your window or on your television. \
You feel it when you go to work, or go to church or pay your taxes.'

text2 = 'You take the blue pill and the story ends.  You wake in your bed and you believe whatever you want to believe \
You take the red pill and you stay in Wonderland and I show you how deep the rabbit-hole goes.'

texts = [text1, text2]
texts
```

```python
# countVEctorizer로 BOW 벡터화
from sklearn.feature_extraction.text import CountVectorizer # BOW(단어 빈도) 벡터화 도구

count_vectorizer = CountVectorizer() # CountVectorizer 객체 생성
count_vectorizer.fit(texts) # 단어 사전 학습 (vocabulary 생성)
text_vecs = count_vectorizer.transform(texts) # 단어를 BOW 벡터화 변환 (희소행렬 생성)

print(text_vecs)

print(type(text_vecs), type(text_vecs.toarray()))   # <class 'scipy.sparse._csr.csr_matrix'>
text_vecs.toarray() # 희소행렬을 배열 형태로 변환
```

```python
print(count_vectorizer.get_feature_names_out())  # 단어 사전 (특성 이름 목록)
print(count_vectorizer.vocabulary_) # 단어 사전 확인
```

```python
import pandas as pd

vacab = sorted(count_vectorizer.vocabulary_.items(), key=lambda x:x[1]) # 인덱스 기준 오름차순 정렬
vocab_df = pd.DataFrame(vacab, columns=['word', 'idx']) # (단어,인덱스) 데이터프레임 생성
vocab_df
```

```python
# 단어 빈도수 계산하여 DataFrame에 추가
word_counts = text_vecs.toarray().sum(axis=0) # 전체 문서 기준 단어별 합계

print(word_counts)

vocab_df['count'] = vocab_df['idx'].apply(lambda i: word_counts[i]) # 빈도값 컬럼 추가

vocab_df = vocab_df.drop(columns=['idx']) 

vocab_df
```

```python
# Countvectorizer 불용어 제거 및 벡터화
count_vectorizer = CountVectorizer(stop_words='english') # 영어 불용어 제거 CountVectorizer 객체 생성
texts_vecs = count_vectorizer.fit_transform(texts) # 단어 사전 학습(생성) 및 벡터화 적용
print(texts_vecs.toarray().shape)

vacab = sorted(count_vectorizer.vocabulary_.items(), key=lambda x:x[1]) # 인덱스 기준 오름차순 정렬
vocab_df = pd.DataFrame(vacab, columns=['word', 'idx']) # (단어,인덱스) 데이터프레임 생성
vocab_df
```

```python
count_vectorizer = CountVectorizer(
    stop_words='english', # 영어 불용어 적용
    ngram_range=(1, 2),   # 1-gram(단어)과 2-gram(바이그램) 모두 추출
    max_features=20       # 빈도수 기준 상위 20개 특성만 사용
)

texts_vecs = count_vectorizer.fit_transform(texts) # 학습 및 BOW변환(희소행렬)
print(texts_vecs.toarray().shape) # (문서수,특성수)

count_vectorizer.get_feature_names_out() # 단어 사전 확인(선택된 단어/바이어그램)
```

**TfidfVectorizer의 주요 파라미터**

  </tr>
  <tr style="background-color: #ffeb99;">
    <td><b>min_df</b></td>

  </tr>

    <td><b>ngram_range</b></td>

    <td>(1, 1)</td>
  </tr>
  <tr style="background-color: #ffeb99;">
    <td>stop_words</td>
    <td>불용어를 지정할 수 있다. "english"로 설정하면 영어 불용어를 사용한다.</td>
    <td>None</td>
  </tr>
  <tr>
    <td>max_features</td>
    <td>벡터화할 때 고려할 최대 단어 수를 설정한다. 빈도순으로 상위 단어들이 선택된다.</td>
    <td>None</td>
  </tr>
  <tr>
    <td>use_idf</td>
    <td>IDF(역문서 빈도)를 사용할지 여부를 지정한다. False로 설정하면 단순히 TF 값만 사용한다.</td>
    <td>True</td>
  </tr>
  <tr>
    <td>smooth_idf</td>

    <td>True</td>
  </tr>

    <td>TF 값에 대해 sublinear scaling (1 + log(tf))를 적용할지 지정한다.</td>

  </tr>
